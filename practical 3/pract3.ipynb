```python
import pandas as pd
import numpy as np
df = pd.read_csv("housing - housing.csv")
# Select numerical variables
numerical_data = df.select_dtypes(include=np.number)
# Calculate correlation matrix
correlation_matrix = numerical_data.corr()
print("Complete Correlation Matrix:")
print(correlation_matrix)
# Get correlation pairs
pairs = []
columns = correlation_matrix.columns
for i in range(len(columns)):
    for j in range(i + 1, len(columns)):
        pairs.append(
            (columns[i], columns[j], correlation_matrix.iloc[i, j])
        )
# Strongest positive correlation
strongest_positive = max(pairs, key=lambda x: x[2])
# Strongest negative correlation
strongest_negative = min(pairs, key=lambda x: x[2])
# Weakest correlation
weakest = min(pairs, key=lambda x: abs(x[2]))
print("\nStrongest Positive Correlation:")
print(strongest_positive)
print("\nStrongest Negative Correlation:")
print(strongest_negative)
print("\nWeakest Correlation:")
print(weakest)
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error
df = pd.read_csv("housing - housing.csv")
# Independent variables
X = df[
    [
        "median_income",
        "housing_median_age",
        "total_rooms",
        "population",
        "households"
    ]
]
# Dependent variable
y = df["median_house_value"]
# Multiple Linear Regression
model = LinearRegression()
model.fit(X, y)
# Predictions
y_pred = model.predict(X)
# Evaluation
r2 = r2_score(y, y_pred)
mse = mean_squared_error(y, y_pred)
print("Multiple Linear Regression")
print("--------------------------")
print("Intercept:", model.intercept_)
print("\nCoefficients:")
for feature, coefficient in zip(X.columns, model.coef_):
    print(feature, ":", coefficient)
print("\nR2 Score:", r2)
print("Mean Squared Error:", mse)
```

```python
# Simple Linear Regression
X_simple = df[["median_income"]]
simple_model = LinearRegression()
simple_model.fit(X_simple, y)
# Predictions
y_pred_simple = simple_model.predict(X_simple)
# Evaluation
r2_simple = r2_score(y, y_pred_simple)
mse_simple = mean_squared_error(y, y_pred_simple)
print("Simple Linear Regression")
print("------------------------")
print("R2 Score:", r2_simple)
print("Mean Squared Error:", mse_simple)
```
