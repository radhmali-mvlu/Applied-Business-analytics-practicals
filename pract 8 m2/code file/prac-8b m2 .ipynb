```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("supermarket_sales.csv")
```

```python
# Select Total column as Revenue
revenue = df["Total"].dropna()
```

```python
# 16. Calculate original sample median
original_median = revenue.median()
print("Original Sample Median:", original_median)
```

```python
# 17 Create one bootstrap sample
bootstrap_sample = np.random.choice(
    revenue,
    size=len(revenue),
    replace=True
)
print("Bootstrap Sample:")
print(bootstrap_sample)
```

```python
# 18 Calculate median
bootstrap_median = np.median(bootstrap_sample)
print("Median of Bootstrap Sample:", bootstrap_median)
```

```python
# 19 Repeat bootstrapping 1000 times
bootstrap_medians = []
for i in range(1000):
    sample = np.random.choice(
        revenue,
        size=len(revenue),
        replace=True
    )
    bootstrap_medians.append(np.median(sample))
print("Total Bootstrap Medians:", len(bootstrap_medians))
```

```python
# 20. Histogram of bootstrap median values
plt.hist(bootstrap_medians, bins=30)
plt.xlabel("Bootstrap Median")
plt.ylabel("Frequency")
plt.title("Bootstrap Distribution of Median Revenue")
plt.show()
```

```python
# 21. Calculate percentiles
lower = np.percentile(bootstrap_medians, 2.5)
upper = np.percentile(bootstrap_medians, 97.5)

print("2.5th Percentile:", lower)
print("97.5th Percentile:", upper)
```

```python
# 22. 95% Bootstrap Confidence Interval
print("95% Bootstrap Confidence Interval:")
print("Lower:", lower)
print("Upper:", upper)
```

```python
# 23. Display original median and confidence interval
print("\nOriginal Sample Median:", original_median)
print("95% Confidence Interval:", (lower, upper))
```

```python
# 24. Compare original median with confidence interval
if lower <= original_median <= upper:
    print("The original median lies within the confidence interval.")
else:
    print("The original median lies outside the confidence interval.")
```

```python
# 25. Interpretation
print("\nInterpretation:")
print(
    "The population median supermarket revenue is reasonably "
    "estimated to lie between", lower, "and", upper
)
```
