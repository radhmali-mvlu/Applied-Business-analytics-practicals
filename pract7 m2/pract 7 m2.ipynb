```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

# -----------------------------------
# PART A: SURVEY AND SAMPLING
# -----------------------------------


```python
# 1. Read CSV file
df = pd.read_csv("Bird Rescue Servey .csv")

# 2. Display first five records
print("FIRST FIVE RECORDS:")
print(df.head())

# 3. Population and Sample
print("\nTotal Population/Responses:", len(df))

# Select numerical variable
# Q6 = How many times have you seen an animal or bird rescue post...
data = pd.to_numeric(df.iloc[:, 8], errors="coerce").dropna()

print("Valid observations:", len(data))
# 4. Random sample of 30 observations
sample = data.sample(n=30, random_state=42)

# 5. Sample statistics
sample_size = len(sample)
sample_mean = sample.mean()
sample_median = sample.median()
sample_std = sample.std()

print("\nPART A - SAMPLE STATISTICS")
print("Sample Size:", sample_size)
print("Sample Mean:", sample_mean)
print("Sample Median:", sample_median)
print("Sample Standard Deviation:", sample_std)

# 6. Chart of collected numerical data
plt.figure(figsize=(7, 5))
plt.hist(data, bins=8, edgecolor="black")
plt.title("Distribution of Rescue Posts Seen")
plt.xlabel("Number of Rescue Posts")
plt.ylabel("Frequency")
plt.show()



```

# -----------------------------------
# PART B: SAMPLING DISTRIBUTION & CLT
# -----------------------------------

```python
sample_size = 30
number_of_samples = 100
sample_means = []

# Repeat random sampling 100 times
for i in range(number_of_samples):

    random_sample = data.sample(
        n=sample_size,
        replace=True
    )

    mean = random_sample.mean()
    sample_means.append(mean)

# Display first 10 sample means
print("\nFIRST 10 SAMPLE MEANS:")

for i in range(10):
    print("Sample", i + 1, "Mean =", sample_means[i])

# Mean of all sample means
mean_of_sample_means = np.mean(sample_means)

print("\nMean of All Sample Means:",
      mean_of_sample_means)

# Standard deviation of sample means
std_sampling_distribution = np.std(
    sample_means,
    ddof=1
)

print("Standard Deviation of Sample Means:",
      std_sampling_distribution)

# Standard Error
standard_error = data.std(ddof=1) / np.sqrt(sample_size)

print("Standard Error:",
      standard_error)

# Original population mean
population_mean = data.mean()

print("\nPopulation Mean:",
      population_mean)

# Histogram of sampling distribution
plt.figure(figsize=(8, 5))

plt.hist(
    sample_means,
    bins=10,
    edgecolor="black"
)

plt.title("Sampling Distribution of Sample Mean")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")

plt.show()

```

```python
# -----------------------------------
# COMPARISON
# -----------------------------------

print("\nCOMPARISON")
print("Original Sample Mean:", sample_mean)
print("Mean of Sampling Distribution:",
      mean_of_sample_means)

print("\nPopulation Mean:", population_mean)

print("\nCONCLUSION:")
print("The mean of the sampling distribution is close")
print("to the population mean, demonstrating the")
print("Central Limit Theorem.")
```

```python
# ==========================================
# POINT 24: DIFFERENT SAMPLE SIZES
# ==========================================

sample_sizes = [5, 15, 30]

plt.figure(figsize=(10, 6))

for size in sample_sizes:
    means = []

    for i in range(100):
        random_sample = data.sample(n=size, replace=True)
        means.append(random_sample.mean())

    plt.hist(means, bins=10, alpha=0.5, label=f"Sample Size = {size}")

plt.title("Sampling Distributions for Different Sample Sizes")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

```python
# ==========================================
# POINT 25: STANDARD ERROR
# ==========================================

standard_error = data.std(ddof=1) / np.sqrt(30)

print("\nPOINT 25 - STANDARD ERROR")
print("Standard Error of Sample Mean:", standard_error)
```

```python
# ==========================================
# POINT 27: ESTIMATE POPULATION MEAN
# ==========================================

print("\nPOINT 27 - ESTIMATION OF POPULATION MEAN")

estimated_population_mean = sample_mean

print("Sample Mean:", sample_mean)
print("Estimated Population Mean:", estimated_population_mean)
```
