```python
#Import Library
from scipy.optimize import linprog
```

```python
#Define Decision Variables
# x = units of Product A
# y = units of Product B
```

```python
#Define Objective Function
c = [-40, -30]
```

```python
#Machine-Hour Constraint
machine = [2, 1]
```

```python
#Labour-Hour Constraint
labour = [1, 2]
```

```python
#Non-Negative Constraints
bounds = [(0, None), (0, None)]
```

```python
#Solve the Linear Programming Problem
result = linprog(
    c,
    A_ub=[
        [2, 1],
        [1, 2]
    ],
    b_ub=[100, 80],
    bounds=bounds,
    method="highs"
)

print(result)
```

```python
#Find Optimal Units
A = result.x[0]
B = result.x[1]

print("Product A:", A)
print("Product B:", B)
```

```python
#Calculate Maximum Profit
profit = 40 * A + 30 * B

print("Maximum Profit:", profit)
```

```python
#Check Available Resources
machine_used = 2 * A + B
labour_used = A + 2 * B

print("Machine Hours Used:", machine_used)
print("Labour Hours Used:", labour_used)
```

```python
#Display Optimal Solution
print("\nOptimal Solution")
print("----------------------")
print("Product A:", A, "units")
print("Product B:", B, "units")
print("Maximum Profit: ₹", profit)
print("Machine Hours Used:", machine_used)
print("Labour Hours Used:", labour_used)
```

```python
#Interpretation
#The company should produce 40 units of Product A and 20 units of Product B.
#  This production plan uses the available machine and labour hours and gives a maximum profit of ₹2200.
```

```python
#Conclusion

#The optimization model determines the best production plan.
#The company should produce 40 units of Product A and 20 units of Product B, giving a maximum profit of ₹2200 while satisfying the available machine and labour hour constraints.
```

```python
#Generate Random Daily Demand for 30 Days
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)

demand = np.random.randint(10, 51, 30)

print("Daily Demand:")
print(demand)
```

```python
#Display Generated Demand Values
print("Demand for 30 Days:")
for day, d in enumerate(demand, 1):
    print("Day", day, ":", d, "units")
```

```python

#Select Inventory Level of 30 Units
inventory = 30
print("Inventory Level:", inventory)
```

```python
#Compare Daily Demand with Inventory
for day, d in enumerate(demand, 1):
    if d <= inventory:
        print("Day", day, ": Demand =", d, "Inventory =", inventory)
    else:
        print("Day", day, ": Demand =", d, "Inventory =", inventory, "Shortage")
```

```python
#Calculate Units Sold Each Day
units_sold = np.minimum(demand, inventory)

print("Units Sold Each Day:")
print(units_sold)
```

```python
#Calculate Shortage
shortage = np.maximum(demand - inventory, 0)

print("Shortage Each Day:")
print(shortage)
```

```python
#Calculate Total Units Sold and Total Shortage
total_sold = units_sold.sum()
total_shortage = shortage.sum()

print("Total Units Sold:", total_sold)
print("Total Shortage:", total_shortage)
```

```python
#Repeat Simulation for 30, 40 and 50 Units
inventory_levels = [30, 40, 50]

results = []

for inventory in inventory_levels:
    units_sold = np.minimum(demand, inventory)
    shortage = np.maximum(demand - inventory, 0)

    total_sold = units_sold.sum()
    total_shortage = shortage.sum()

    results.append([
        inventory,
        total_sold,
        total_shortage
    ])

print(results)
```

```python
#Store Results in DataFrame
df = pd.DataFrame(
    results,
    columns=["Inventory Level", "Total Sales", "Total Shortage"]
)

print(df)
```

```python
#Compare Total Sales and Shortage
print("Comparison of Inventory Levels:")
print(df)
```

```python
#Create Chart
plt.bar(df["Inventory Level"] - 2, df["Total Sales"],
        width=4, label="Total Sales")

plt.bar(df["Inventory Level"] + 2, df["Total Shortage"],
        width=4, label="Total Shortage")

plt.xlabel("Inventory Level")
plt.ylabel("Units")
plt.title("Inventory Level vs Sales and Shortage")
plt.legend()
plt.show()
```

```python
#Identify Effect of Inventory Level
print("Effect of Inventory Level:")
print("As inventory level increases, total sales increase or remain stable.")
print("As inventory level increases, total shortage decreases.")
```

```python
#Conclusion
#The simulation shows that increasing the inventory level increases total sales and reduces shortage.
#A suitable inventory level helps the supermarket meet customer demand while minimizing stock shortages.
```
