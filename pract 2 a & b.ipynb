```python
import pandas as pd
df = pd.read_csv("supermarket_sales.csv")
print(df.head())
```

```python
sales = df["Total"]
print("Mean :", sales.mean())
print("Median :", sales.median())
print("Mode :")
print(sales.mode())
print("Minimum :", sales.min())
print("Maximum :", sales.max())
print("Range :", sales.max() - sales.min())
print("Variance :", sales.var())
print("Standard Deviation :", sales.std())
print("Count :", sales.count())
print("Sum :", sales.sum())
print("Descriptive Statistics")
print(sales.describe())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))
sns.histplot(df["Total"], bins=10, kde=False, color="skyblue")
plt.title("Histogram of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df["Total"], color="orange")
plt.title("Box Plot of Sales")
plt.xlabel("Sales")
plt.show()
```
